# SQL Analysis

This notebook performs SQL analysis on the Gold layer tables.

The analysis provides business insights from the electricity consumption data.

Tables Used:
- gold_hourly_consumption
- gold_daily_consumption
- gold_monthly_consumption
- gold_prediction
- gold_anomaly_alerts

```python
%sql

SELECT *
FROM workspace.default.gold_daily_consumption;
```

```python
%sql

SELECT
    meter_id,
    reading_date,
    total_daily_units
FROM workspace.default.gold_daily_consumption
ORDER BY total_daily_units DESC
LIMIT 10;
```

```python
%sql

SELECT
    meter_id,
    total_monthly_units
FROM workspace.default.gold_monthly_consumption
ORDER BY total_monthly_units DESC;
```

```python
%sql

SELECT
    AVG(total_monthly_units) AS average_monthly_consumption
FROM workspace.default.gold_monthly_consumption;
```

```python
%sql

SELECT
    meter_id,
    reading_date,
    sma_7day
FROM workspace.default.gold_prediction
ORDER BY sma_7day DESC
LIMIT 10;
```

```python
%sql

SELECT
    anomaly_type,
    COUNT(*) AS total_records
FROM workspace.default.gold_anomaly_alerts
GROUP BY anomaly_type;
```

```python
%sql

SELECT *
FROM workspace.default.gold_anomaly_alerts
WHERE severity = 'HIGH';
```

```python
%sql

SELECT
    severity,
    COUNT(*) AS total_alerts
FROM workspace.default.gold_anomaly_alerts
GROUP BY severity;
```

```python
%sql

SELECT
    meter_id,
    COUNT(*) AS alert_count
FROM workspace.default.gold_anomaly_alerts
WHERE anomaly_type <> 'NORMAL'
GROUP BY meter_id
ORDER BY alert_count DESC;
```

```python
%sql

SELECT 'gold_hourly_consumption' AS table_name,
COUNT(*) AS total_rows
FROM workspace.default.gold_hourly_consumption

UNION ALL

SELECT 'gold_daily_consumption',
COUNT(*)
FROM workspace.default.gold_daily_consumption

UNION ALL

SELECT 'gold_monthly_consumption',
COUNT(*)
FROM workspace.default.gold_monthly_consumption

UNION ALL

SELECT 'gold_prediction',
COUNT(*)
FROM workspace.default.gold_prediction

UNION ALL

SELECT 'gold_anomaly_alerts',
COUNT(*)
FROM workspace.default.gold_anomaly_alerts;
```
